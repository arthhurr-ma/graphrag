```python
import os
import shutil
import random
```

```python
def get_all_files_in_directory(directory):
    """Helper function to get all files (recursively) from a directory."""
    all_files = []
    for root, _, files in os.walk(directory):
        for file in files:
            all_files.append(os.path.join(root, file))
    return all_files
```

```python
def find_file(base_name, search_directory):
    for root, dirs, files in os.walk(search_directory):
        if base_name in files:
            relative_path = os.path.relpath(os.path.join(root, base_name), start=search_directory)
            return relative_path
    return None
```

```python

def copy_random_files(storage_dir, input_dir, num_files_to_copy):

    folder_mapping = {
        '10Q': '10q',
        '10K': '10k',
        'new_earnings_call': 'earnings call',
        'old_earnings_call': 'earnings call',
        'old_rating_report': 'ratings report',
        'new_rating_report': 'ratings report'
    }

    input_files = set(os.path.basename(file) for file in get_all_files_in_directory(input_dir))
    print(len(input_files))

    storage_files = set(os.path.basename(file) for file in get_all_files_in_directory(storage_dir))
    print(len(storage_files))
    
    available_files = [file for file in storage_files if os.path.basename(file) not in input_files]
    print(f"Available files to copy: {len(available_files)}")


    num_files_to_copy = min(num_files_to_copy, len(available_files))
    selected_files = random.sample(available_files, num_files_to_copy)

    selected_files_with_storage_subdir = [find_file(file, storage_dir) for file in selected_files]

    selected_files_with_input_subdir = []
    for file_with_storage_subdir in selected_files_with_storage_subdir:
        mapped_path = file_with_storage_subdir
        for old, new in folder_mapping.items():
            mapped_path = mapped_path.replace(old, new)
        selected_files_with_input_subdir.append(mapped_path)

    
    for storage_file, input_file in zip(selected_files_with_storage_subdir, selected_files_with_input_subdir):
    
        input_file_path = os.path.join(input_dir, input_file)
        storage_file_path = os.path.join(storage_dir, storage_file)
        print(f"Copying {storage_file_path} to {input_file_path}")
        input_file_dir = os.path.dirname(input_file_path)
        os.makedirs(input_file_dir, exist_ok=True)
        shutil.copy(storage_file_path, input_file_path)

    
 

```

```python
output_dir = "/Users/arthurj/Git_Repos/graphrag copy/graphrag_moodys_v2.0.0_test2/input/"
storage_dir = "/Users/arthurj/Git_Repos/graphrag copy/graphrag/extra_storage/"

copy_random_files(storage_dir, output_dir, 100)
```

```python
def check_files_in_folder(folder_path):
    total_valid_files = 0

    for subfolder in os.listdir(folder_path):
        subfolder_path = os.path.join(folder_path, subfolder)
        if os.path.isdir(subfolder_path):
            valid_file_count = 0
        else:
            continue

        for file in os.listdir(subfolder_path):
            file_path = os.path.join(subfolder_path, file)
            if os.path.isfile(file_path):
                try:
                    with open(file_path, 'r') as file:
                        lines = file.readlines()
                        if len(lines) >= 5:
                            valid_file_count += 1
                except Exception as e:
                    print(f"Error reading file {file_path}: {e}")

        print(f"Number of valid files in {subfolder} is {valid_file_count}")
        total_valid_files += valid_file_count

    print(f"Total number of valid files: {total_valid_files}")

output_dir = "/Users/arthurj/Git_Repos/graphrag copy/graphrag_moodys_v2.0.0_test2/input/"
check_files_in_folder(output_dir)
```
